```python
def parse_speech_file(filepath, corpus_type):
    """
    Parse a UN General Debate speech file named like:
    AFG_71_2016.txt

    Returns one dictionary representing one speech.
    """

    stem = filepath.stem
    parts = stem.split("_")

    if len(parts) != 3:
        raise ValueError(f"Unexpected filename format: {filepath.name}")

    iso3 = parts[0]
    session = int(parts[1])
    year = int(parts[2])

    text = filepath.read_text(
        encoding="utf-8",
        errors="replace"
    )

    return {
        "corpus": corpus_type,
        "iso3": iso3,
        "session": session,
        "year": year,
        "filename": filepath.name,
        "filepath": str(filepath),
        "text": text,
        "word_count": len(text.split())
    }
```

```python
focal_files = sorted(FOCAL_DIR.rglob("*.txt"))
dev_files = sorted(DEV_DIR.rglob("*.txt"))
```

```python
focal_records = [
    parse_speech_file(f, "focal")
    for f in focal_files
]

focal_df = pd.DataFrame(focal_records)
```

```python
FOCAL_METADATA = {
    "USA": ("United States", "Modern West"),
    "GBR": ("United Kingdom", "Modern West"),

    "CHN": ("China", "East Asia"),
    "JPN": ("Japan", "East Asia"),

    "SAU": ("Saudi Arabia", "Arab region"),
    "JOR": ("Jordan", "Arab region"),

    "COL": ("Colombia", "Latin America"),
    "CHL": ("Chile", "Latin America"),

    "IND": ("India", "South Asia"),
    "PAK": ("Pakistan", "South Asia"),

    "GHA": ("Ghana", "Sub-Saharan Africa"),
    "KEN": ("Kenya", "Sub-Saharan Africa"),

    "IDN": ("Indonesia", "Southeast Asia"),
    "THA": ("Thailand", "Southeast Asia"),

    "KAZ": ("Kazakhstan", "Central Asia"),
    "TJK": ("Tajikistan", "Central Asia"),

    "POL": ("Poland", "Eastern Europe"),
    "ROU": ("Romania", "Eastern Europe")
}
```

```python
focal_df["country"] = focal_df["iso3"].map(
    lambda x: FOCAL_METADATA[x][0]
)

focal_df["region"] = focal_df["iso3"].map(
    lambda x: FOCAL_METADATA[x][1]
)
```
